```python
factors = [27, 34, 22, 11, 4, 2] # combine 11, 4, and 2

def get_partitions(num_cards):
    p = [[num_cards]]
    for i in range(1, num_cards):
        main = [i, num_cards-i]
        main.sort()
        if not main in p:
            p.append(main)
            parts = get_partitions(num_cards-i)
            for sub in parts:
                cand = [i]
                for pp in sub:
                    cand.append(pp)
                cand.sort()
                if not cand in p:
                    p.append(cand)
    return p

get_partitions(7)
```

```python
factors_sum = [0]
for f in factors:
    factors_sum.append(factors_sum[-1] + f)

factors_sum = factors_sum[:4] + [100]
factors_sum
```

```python
import random

num_cards = 5
parts = get_partitions(num_cards)
parts_hist = [0]*len(parts)
n_trials = 10000

for _ in range(n_trials):
    hst = [0, 0, 0, 0]
    for c in range(num_cards):
        n = random.randint(0,99)
        for i,fs in enumerate(factors_sum):
            if fs > n:
                hst[i-1] += 1
                break
    hst2 = []
    for h in hst:
        if h > 0:
            hst2.append(h)
    hst = hst2
    hst.sort()
    s = sum(hst)
    for i in range(num_cards-s):
        hst = [1] + hst
    for i,hand in enumerate(parts):
        if hst == hand:
            parts_hist[i] += 1
            break

zp = list(zip(parts, parts_hist))
zp.sort(key=lambda a: a[1])

for p,h in zp:
    if len(p) > 4:
        continue
    print(p, f"{h/n_trials*100:.2f}%")
        
```
